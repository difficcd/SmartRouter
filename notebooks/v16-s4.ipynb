```python
# v16 안전성 탐색 -- 슬라이스 s4: balanced, risk_high=3.0, 4.0, 5.0, 6.0 (24조합)
# 편집할 것 없음. [런타임] > [모두 실행] 만 누르면 된다.
# 끝나면 BEGINs4 / ENDs4 사이의 문자열을 통째로 복사해서 돌려주면 된다.
!rm -rf /content/sr && git clone -q --branch SmartRouter-v16 --depth 1     https://github.com/difficcd/SmartRouter.git /content/sr
!cd /content/sr && python training/search_standalone.py     --matrices training/tmp-v16-matrices.npz     --tiers balanced --risk-high 3.0 4.0 5.0 6.0     --workers 2 --out /content/out-s4.json --curve-out /content/curve-s4.json

import base64, gzip, hashlib, json, time
raw = open('/content/curve-s4.json', 'rb').read()
rows = json.loads(raw)['rows']
blob = base64.b64encode(gzip.compress(raw, 9)).decode()
assert len(rows) == 24, '조합 수가 24가 아니다: %d' % len(rows)
print('=' * 30)
print('SLICE=s4 TIER=balanced JOBS=%d PTS=%d SHA=%s DONE_AT=%s' % (
    len(rows), sum(len(r['curve']) for r in rows),
    hashlib.sha256(raw).hexdigest()[:12], time.strftime('%Y-%m-%d %H:%M:%S')))
print('BEGINs4')
print(blob)
print('ENDs4')
```
